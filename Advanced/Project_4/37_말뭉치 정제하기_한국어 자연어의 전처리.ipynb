## 정제 : 정규표현식(re), 글자 반복 제거

```python
#정규표현식 -> 특정한 패턴의 자연어를 찾을 수 있다.
#ㄱ-ㅎ : 한국어 자모의 처음-끝
#가-힣 : 한국어 글자의 처음-끝
#\s : 공백 -> 스페이스바, 한줄띄우기기
# ^ : ~(not)
data = re.sub(r'[^ㄱ-ㅎ가-힣\s]', '', data)
```

```python
data
```

```python
from soynlp.normalizer import emoticon_normalize, repeat_normalize

print(emoticon_normalize('앜ㅋㅋㅋㅋㅋㅋㅋ 진짴ㅋㅋㅋㅋㅋ 너무 웃기닼ㅋㅋㅋ 다시 말해봐밬ㅋㅋㅋㅜㅠㅠㅠㅠ',
                        num_repeats=2))
```

```python
corpus = emoticon_normalize(data, num_repeats=2)
print(corpus)
```

```python
from soynlp.tokenizer import RegexTokenizer

tokenizer = RegexTokenizer()
data = tokenizer.tokenize(corpus)
print(tokenizer.tokenize(corpus))
```

```python
#다운로드 받은 파일을 읽어들어온 다음, \n을 기준으로 나눠서
#stop_words라는 변수에 저장
with open('./korean_stopwords.txt', 'r', encoding='utf-8') as f:
  stop_words = f.read().split("\n")

print(len(stop_words), stop_words[:10])
```

```python
#내가 가진 text라는 문자를 '불용어 제거' 하는 코드
result=[]
for i in data:
  if i not in stop_words:
    result.append(i)
```

### 스테밍(어간 추출), 레마타이즈(표제어 추출)

```python
from soylemma import Lemmatizer

lemmatizer = Lemmatizer()
lemmatizer.analyze('차가우니까')
```

```python
# 간단한 규칙 기반 토크나이저 (어미 제거)
def simple_stem_tokenizer(text):
    # 자주 사용되는 어미 리스트 (원형 복원 목적)
    endings = ['습니다', '었다', '했다', '해요', '하는']
    stemmed = []
    for word in text:
        for end in endings:
            if word.endswith(end):       # 어미로 끝나면
                word = word[:-len(end)]  # 어미 제거
        stemmed.append(word)
    return stemmed
```

```python
removed = []
for i in result:
    removed.append(simple_stem_tokenizer(result))
```

```python
from soynlp.noun import LRNounExtractor
noun_extractor = LRNounExtractor()
nouns = noun_extractor.train_extract(result)
```
