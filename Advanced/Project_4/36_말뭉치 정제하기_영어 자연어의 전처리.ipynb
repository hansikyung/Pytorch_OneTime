#### 컴퓨터는 단어를 알아들을 수 없음 >> 텍스트 데이터의 숫자화(임베딩) >> 효율적인 임베딩을 위해 자연어에 대한 처리를 수행

1. 정제와 정규표현식을 사용
- 원시 텍스트(text)에서 불필요한 요소를 제거하고, 필요한 데이터 형식으로 조정
- 데이터 분석의 정확도와 성능에 영향을 미침

2. 자연어 처리 : 토큰화
- 텍스트를 의미있는 요소인 토큰(단어, 문장 등)으로 분리하는 과정
- 문법적, 의미적 분석의 기본 단위를 설정   
- 기존에 지정된 '토크나이저'를 가져와 사용하는 경우가 많음
ㅡ
3. 자연어 처리 : 불용어 제거
- 분석에 불필요한 정보(예: "그리고", "하지만")를 제거
- 데이터의 노이즈를 줄이고, 처리 과정의 효율성을 높이는 데 도움을 주는 과정

4. 자연어 처리 : 스테밍(어간 추출)
- 단어에서 접미사를 제거하여 기본 형태(어간)을 추출
- 문법적 변형으로 인한 단어의 수를 줄여, 단어의 일반적인 의미를 파악하는 데 도움을 줌

5. 자연어 처리 : 레마타이즈(표제어 추출)
- 단어를 사전형으로 변환하여 정확한 기본형(표제어)을 찾음
- 문맥을 고려하여 단어의 정확한 문법적 정보를 유지하는 데 유리

### 토큰화

```python
#영어 문장을 토큰화하기 위한 라이브러리
!pip install nltk
```

```python
import nltk

#단어모음, 토크나이징 모음...
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
#단어 단위로 끊을 수 있도록 word_tokenize를 import
from nltk.tokenize import word_tokenize

#WordPucntTokenizer를 가져와서 토큰화를 진행
from nltk.tokenize import WordPunctTokenizer
```

```python
print("단어 토큰화 : ")
print(word_tokenize("Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."))
```

```python
print("단어 토큰화 : ")
print(WordPunctTokenizer().tokenize("Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."))
```

### 불용어 제거

```python
nltk.download('stopwords')
```

```python
#영어의 불용어들을 프린트해서 확인
print(nltk.corpus.stopwords.words('english'))
```

```python
#샘플이 될 문장
sentence = "Today is happy day. It's raining outside."

#불용어를 중복이 없는 'set'자료형에 저장
stop_words = set(nltk.corpus.stopwords.words('english'))

#토크나이징이 먼저 > 왜? > 불용어 제거 전에 '단어 단위'로 쪼개야 하기 때문.
word_tokens = word_tokenize(sentence)
print(word_tokens)
```

```python
result = []

for word in word_tokens:
  if word not in stop_words:
    result.append(word)

print('불용어 제거 전 ', word_tokens )
print('불용어 제거 후 ', result)
```

### 표제어 추출

```python
nltk.download('wordnet')
```

```python
#표제어 추출을 도와주는 표제어 추출기
from nltk.stem import WordNetLemmatizer

lemmetizer = WordNetLemmatizer()

words = ['policies', 'doing', 'has', 'going', 'loves', 'jumping']

print("표제어 추출 전 : ", words)
print("표제어 추출 후 : ", [lemmetizer.lemmatize(word) for word in words])
```

```python
#표제어 추출의 정정
#동사 verb
lemmetizer.lemmatize("has", "v")
```

### 어간 추출

```python
#nltk.download('punkt')
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

sentence = "This is the test for the youth. Do you know all of these questions?"

tokenized = word_tokenize(sentence)
print(tokenized)
```

```python
sentence = "Are you student? Please submit your id card to enrollment."

tokenized = word_tokenize(sentence)

print('어간 추출 전 : ', tokenized)
print('어간 추출 후 : ',  [stemmer.stem(w) for w in tokenized])
```
