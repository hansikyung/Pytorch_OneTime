### 19-1. Computer Vision의 Task란

- Computer Vision 기반의 Deep Learning 태스크에는 몇 가지 갈래가 있습니다.
- 1. 이미지 분류 (Image Classification): 이미지를 입력받아 어떤 클래스에 속하는지 예측함.
- 2. 객체 검출 (Object Detection): 이미지 내 객체의 위치(경계 상자)와 클래스를 예측함
- 3. 이미지 분할 (Image Segmentation): 이미지의 각 픽셀이 어떤 클래스에 속하는지 분류함
- 4. 행동 인식 (Action Recognition): 영상에서 사람이나 객체의 행동을 분석하여 분류함
- 5. 이미지 생성 (Image Generation): 새로운 이미지를 생성하거나 기존 이미지를 변환함

- 이 외에도 여러가지 분야에서의 Task가 있으나, 처음 Computer Vision에 입문하는 경우 만나게 되는 주제는 주로 위의 5가지 중 하나가 될 것입니다.
- 우리는 이 중에서 2번, 객체 검출(Object Detection)에 대해 프로젝트를 진행하게 됩니다.

### 19-2. Object Detection 이란       

- Computer Vision 기반의 Task에서, 이미지 분류, 객체 검출, 이미지 분할은 각각 같지만 다른 태스크를 수행합니다.
- 이미지 분류는 데이터셋에 속한 한 장의 이미지를 보고, 이미지 안의 대상이 무엇인지를 분류합니다.
- 객체 분류는 데이터셋에 속한 한 장의 이미지를 보고, 이미지 안에서 대상을 찾고, 대상이 무엇인지를 분류합니다.
- 즉, 이미지 분류와 객체 분류 사이에는 1개의 스텝(+이미지 안에서 대상의 위치를 찾는다) 가 추가되는 것입니다.
- 이미지 분할은 데이터셋에 속한 한 장의 이미지를 보고, 이미지 안에서 대상을 찾고, 대상이 무엇인지를 분류하되, 대상의 범위까지 픽셀 단위로 분류합니다.
- 즉, 객체 분류와 이미지 분할 사이에도 1개의 스텝(+대상의 위치를 찾고, 대상의 범위를 픽셀 단위로 분리한다) 가 추가되는 것입니다.     

## 19-3 Object Detection 모델의 데이터셋이란?        

- Object Detection을 수행하는 모델 아키텍쳐는 여러 가지가 있고, 그에 따라 모델에 Input되는 데이터셋도 여러가지 포맷을 가졌습니다.
- 하지만 여기에서, 주로 쓰이는 3가지 유형의 데이터셋 포맷에 대해 살펴보고자 합니다.
- 주요 데이터셋 포맷은 YOLO, COCO, Pascal VOC입니다.
- 물론, 데이터셋에서 '이미지를 활용한다' 라는 것은 같지만 '라벨이 어떻게 주어지느냐' 에 대한 차이가 있다! 라는 점을 기억해주시면 좋겠습니다.   


- YOLO(You Only Look Once) 모델에 최적화된 라벨의 모양은 아래와와 같습니다.
- 라벨 구조:
    - 하나의 텍스트 파일에 이미지의 객체 정보를 저장함.
    - 파일 이름은 이미지 이름과 동일하고 확장자가 .txt로 매칭됨.
    - 각 줄은 하나의 객체를 나타냄.
    - 하나의 객체는 0~1 사이로 정규화된 값으로 표현됨 
    - <class_id> <x_center> <y_center> <width> <height>

- COCO 타입 데이터셋 라벨의 모양은 아래와와 같습니다.
- 라벨 구조:
    - JSON 파일로 저장됨.
    - 이미지와 객체 정보, 클래스, 카테고리, 메타데이터들을 모두 포함하고 있음(라벨에 포함된 정보가 많은 편임).
    - 이미지의 위치와 크기는 주로 "bbox": [x_min, y_min, width, height] 의 형태로 표현됨

- VOC 타입 데이터셋 라벨의 모양은 위의 그림과 같습니다.
- 라벨 구조:
    - XML 파일로 저장.
    - 하나의 XML 파일이 하나의 이미지와 연결
    - 이미지의 위치와 크기는 주로 bndbox: 객체의 경계 상자 좌표 (xmin, ymin, xmax, ymax) 의 형태로 표현됨
