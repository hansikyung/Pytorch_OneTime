```python
from ultralytics import YOLO
```

## 모델 가중치 가져오기

```python
#가중치 기반의 모델을 로드하여 추론 결과 확인하기
model = YOLO("yolo8s.pt")
```

```python
# valid set에 대한 결과를 out하여 확인
metrics = model.val()
```

## 테스트 셋에 대한 추론을 수행하기

```python
#path를 입력하여 추론을 수행하고, 이에 대한 결과를 시각화함
results = model("path/to/image.jpg")
results[0].show()
```
